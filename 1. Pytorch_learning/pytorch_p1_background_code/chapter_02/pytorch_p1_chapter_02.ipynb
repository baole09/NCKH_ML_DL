# Chapter 2 — Introduction to PyTorch
PyTorch là gì, cài thế nào, dùng cho việc gì, và PyTorch khác TensorFlow ở đâu?


## 1. Hai năng lực cốt lõi tạo nên PyTorch
1. **Tensor computation** — tính toán trên mảng nhiều chiều, giống NumPy, có thể chạy trên GPU.
2. **Automatic differentiation (autograd)** — tính đạo hàm tự động, nền tảng để huấn luyện mạng neural qua backpropagation.

## 3. Ứng dụng thực tế

3 nhóm chính, mỗi nhóm gắn với model cụ thể:
- Computer Vision: ResNet, YOLO, Faster R-CNN.
- NLP: BERT, Transformer.
- Reinforcement Learning: DQN, PPO, A3C.

## 4. So với TensorFlow/Keras

Lí do **tại sao** hai thư viện khác nhau chứ không chỉ liệt kê khác biệt:

| Khía cạnh | PyTorch | TensorFlow/Keras |
|---|---|---|
| Training loop | Tự viết (`loss.backward()`, `optimizer.step()`) | Có sẵn qua `model.fit()` |
| Triết lý | Kiểm soát chi tiết từng bước — phù hợp nghiên cứu | Ẩn chi tiết, giống scikit-learn — phù hợp ứng dụng nhanh |
| Placeholder tensor | Không có — tensor luôn mang giá trị thật | Keras tensor là placeholder, có lúc gây nhầm lẫn không biết tensor có giá trị hay chưa |
| Visualize model | Không có hàm dựng sẵn như `plot_model()` | Có |
| Thư viện đi kèm | `torchvision` (thị giác máy tính) | Hỗ trợ mobile/web mạnh (TF Lite, TF.js), có TensorBoard |

Kết luận: sự khác biệt nằm ở **mức độ kiểm soát**. PyTorch bắt người dùng viết tường minh mọi bước của training loop, đổi lại linh hoạt hơn khi thử nghiệm ý tưởng mới. Keras giấu các bước đó sau `fit()`, đổi lại nhanh hơn cho các bài toán chuẩn.

## 5. Vận dụng


````python
import torch

# create a tensor of size (2,3)
x = torch.tensor([[1, 2, 3], [4, 5, 6]])

# create a tensor of size (3,2)
y = torch.tensor([[1, 2], [3, 4], [5, 6]])

# matrix multiplication of x and y
z = torch.matmul(x, y)

print(z)
````

## 6. Kết luận

- PyTorch = tensor computation + autograd.
- Cài đặt PyTorch cho GPU đòi hỏi CUDA riêng.
- Sự khác biệt PyTorch vs Keras không nằm ở việc "cái nào tốt hơn", mà ở mức độ kiểm soát người dùng muốn có với training loop.
