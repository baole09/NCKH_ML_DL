# Keras/TensorFlow — Chapter 7: Develop Your First Neural Network with Keras


## 1. Bài toán và dữ liệu

- Dùng bộ dữ liệu **Pima Indians Diabetes** (8 biến đầu vào, phân loại nhị phân). Nạp bằng `numpy.loadtxt()`, tách `X = dataset[:,0:8]`, `y = dataset[:,8]`.

## 2. Định nghĩa model

- Kiến trúc giống hệt bản PyTorch: 8 → 12 (ReLU) → 8 (ReLU) → 1 (Sigmoid).
- Cách viết Keras: `model.add(Dense(12, input_shape=(8,), activation='relu'))` làm **2 việc cùng lúc**: vừa khai báo input layer (qua `input_shape`), vừa là hidden layer đầu tiên.
## 3. Compile, fit, evaluate

- `compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])`.
- `fit(X, y, epochs=150, batch_size=10)`.
- `evaluate(X, y)` trả về `[loss, accuracy]`.
- Gói toàn bộ training loop vào 1 lệnh `fit()`.

## 4. Đánh giá

Sách **chủ động thừa nhận rõ ràng**: `evaluate(X, y)` ở đây dùng lại chính tập đã train, "chỉ cho biết mô hình khớp dữ liệu train tốt đến đâu, không nói được gì về dữ liệu mới" — làm vậy "vì đơn giản", lý tưởng ra nên tách riêng tập train/test.

- **Kết quả**: sau 150 epoch, accuracy trên train ~77.86%. Chạy lại 5 lần cho accuracy dao động 75.00%–78.12%, trung bình 76.924%.

## 5. Dự đoán

- `model.predict(X)` trả xác suất (do output sigmoid) → làm tròn bằng `round()` hoặc gọn hơn `(model.predict(X) > 0.5).astype(int)`.


## 6. Vận dụng


**7.1–7.7** — load data → define → compile → fit → evaluate

```python
# first neural network with keras tutorial
from numpy import loadtxt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
# load the dataset
dataset = loadtxt('pima-indians-diabetes.csv', delimiter=',')
# split into input (X) and output (y) variables
X = dataset[:,0:8]
y = dataset[:,8]
# define the keras model
model = Sequential()
model.add(Dense(12, input_shape=(8,), activation='relu'))
model.add(Dense(8, activation='relu'))
model.add(Dense(1, activation='sigmoid'))
# compile the keras model
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
# fit the keras model on the dataset
model.fit(X, y, epochs=150, batch_size=10)
# evaluate the keras model
_, accuracy = model.evaluate(X, y)
print('Accuracy: %.2f' % (accuracy*100))
```

**7.9–7.11** — huấn luyện + dự đoán và so sánh 5 mẫu đầu

```python
# first neural network with keras make predictions
from numpy import loadtxt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
# load the dataset
dataset = loadtxt('pima-indians-diabetes.csv', delimiter=',')
# split into input (X) and output (y) variables
X = dataset[:,0:8]
y = dataset[:,8]
# define the keras model
model = Sequential()
model.add(Dense(12, input_shape=(8,), activation='relu'))
model.add(Dense(8, activation='relu'))
model.add(Dense(1, activation='sigmoid'))
# compile the keras model
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
# fit the keras model on the dataset
model.fit(X, y, epochs=150, batch_size=10, verbose=0)
# make class predictions with the model
predictions = (model.predict(X) > 0.5).astype(int)
# summarize the first 5 cases
for i in range(5):
    print('%s => %d (expected %d)' % (X[i].tolist(), predictions[i], y[i]))
```
